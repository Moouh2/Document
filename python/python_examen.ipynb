```python
table_titre_Parcs_nationaux = [title.text.strip() for title in titre_Parcs_nationaux]
```

```python
print(table_titre_Parcs_nationaux)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = table_titre_Parcs_nationaux)

df
```

```python
for row in colonne_donnees:
    print(row.find_all('td'))
```

```python
for row in colonne_donnees [1:]:
    row_data = row.find_all('td')
    individuel_ranger_donnees = [data.text.strip() for data in row_data]
    print( individuel_ranger_donnees)
```

```python
for row in colonne_donnees  [1:]:
    row_data = row.find_all('td')
    individuel_ranger_donnees = [data.text.strip() for data in row_data]
    
    length = len(df)
    df.loc[length] = individuel_ranger_donnees
```
