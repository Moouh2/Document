```python
import matplotlib.pyplot as plt
plt.style.use('ggplot')
import seaborn as sns
import pandas as pd

# Créeons un DataFrame avec vos données
data = {
    'Parc national': ["Parc national d'Akanda", "Parc national de Birougou", "Parc national d'Ivindo", "Parc national de Loango", "Parc national de la Lopé"],
    'Année de création': [2002, 2002, 2002, 2002, 2002],
    'Province': ['Estuaire', 'Ngounié, Ogooué-Lolo', 'Ogooué-Ivindo', 'Ogooué-Maritime', 'Ogooué-Ivindo'],
    'Surface (km²)': [411.0, 683.0, 2967.0, 1510.0, 4942.0]
}

df = pd.DataFrame(data)

# Bar plot pour les surfaces par parc
plt.figure(figsize=(10, 6))
sns.barplot(x='Parc national', y='Surface (km²)', data=df)
plt.title('Surfaces par Parc National')
plt.xlabel('Parc National')
plt.ylabel('Surface (km²)')
plt.xticks(rotation=45, ha='right')  # Rotation des noms pour une meilleure lisibilité
plt.show()

```

```python
# Scatter plot pour les années de création et les surfaces par province
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Année de création', y='Surface (km²)', data=df, hue='Province', s=100)
plt.title('Relation entre Année de création et Surfaces par Province')
plt.xlabel('Année de création')
plt.ylabel('Surface (km²)')
plt.legend(title='Province')
plt.show()
```

```python
#Diagramme à secteurs (Pie Chart) pour la distribution des parcs par province :
plt.figure(figsize=(8, 8))
df['Province'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90)
plt.title('Distribution des Parcs par Province')
plt.ylabel('')  # Supprimer le label y
plt.show()
```

```python
#Diagramme en boîte (Box Plot) pour visualiser la répartition des surfaces par province :
plt.figure(figsize=(10, 6))
sns.boxplot(x='Province', y='Surface (km²)', data=df)
plt.title('Répartition des Surfaces par Province')
plt.xlabel('Province')
plt.ylabel('Surface (km²)')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Créeons un tableau croisé dynamique (pivot table) pour la visualisation
pivot_df = df.pivot_table(values='Surface (km²)', index='Année de création', columns='Province', aggfunc='sum')

# Tracer le graphique à barres empilées
plt.figure(figsize=(12, 8))
pivot_df.plot(kind='bar', stacked=True)
plt.title('Distribution des Parcs par Province et Surface selon l\'Année de Création')
plt.xlabel('Année de Création')
plt.ylabel('Surface (km²)')
plt.legend(title='Province', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Compter le nombre de parcs par province
counts_by_province = df['Province'].value_counts()

# Tracer le graphique à barres
plt.figure(figsize=(10, 6))
sns.barplot(x=counts_by_province.index, y=counts_by_province.values)
plt.title('Nombre de Parcs par Province')
plt.xlabel('Province')
plt.ylabel('Nombre de Parcs')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Tracer le graphique à barres groupées
plt.figure(figsize=(12, 8))
sns.countplot(x='Province', hue='Parc national', data=df)
plt.title('Nombre de Parcs par Province avec leurs Noms')
plt.xlabel('Province')
plt.ylabel('Nombre de Parcs')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Parc national', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
