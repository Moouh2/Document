```python
import requests
import sys
## Categorie Chat : 1 - 5016
## Categorie Chien : 5016 - 10032
BASE_URL = "http://localhost:8080"
API_TOKEN = "5803bad81e162f369d0f2d5dc9622a78f1cbba12"
#5803bad81e162f369d0f2d5dc9622a78f1cbba12
# Si "Chat" = "Dadi" (sinon mets "Avu")
LABEL_VALUE = "Avu"

# Noms d'après ta config:
# <Image name="image" value="$image"/>
# <Choices name="choice" toName="image"> ... </Choices>
FROM_NAME = "choice"
TO_NAME = "image"

HEADERS = {
    "Authorization": f"Token {API_TOKEN}",
    "Content-Type": "application/json",
}

def task_exists(task_id: int) -> bool:
    r = requests.get(f"{BASE_URL}/api/tasks/{task_id}/", headers=HEADERS)
    if r.status_code == 404:
        return False
    r.raise_for_status()
    return True

def get_annotation_count(task_id: int) -> int:
    r = requests.get(f"{BASE_URL}/api/tasks/{task_id}/annotations/", headers=HEADERS)
    if r.status_code == 404:
        return 0
    r.raise_for_status()
    js = r.json()
    if isinstance(js, dict) and "count" in js:
        return js["count"]
    if isinstance(js, list):
        return len(js)
    return 0

def create_choice_annotation(task_id: int, value: str = LABEL_VALUE):
    payload = {
        "result": [{
            "from_name": FROM_NAME,
            "to_name": TO_NAME,
            "type": "choices",
            "value": {"choices": [value]},
        }]
    }
    r = requests.post(f"{BASE_URL}/api/tasks/{task_id}/annotations/", headers=HEADERS, json=payload)
    r.raise_for_status()
    return r.json()

def main(start_id, end_id, only_unannotated=True, dry_run=False):
    done = 0
    for tid in range(start_id, end_id + 1):
        # Vérifier que la tâche existe
        if not task_exists(tid):
            print(f"- Task {tid} not found (skip).")
            continue

        # Éviter de réannoter si déjà annotée
        if only_unannotated and get_annotation_count(tid) > 0:
            print(f"- Task {tid} already annotated (skip).")
            continue

        if dry_run:
            print(f"[DRY-RUN] Would annotate task {tid} with '{LABEL_VALUE}'")
        else:
            try:
                create_choice_annotation(tid)
                print(f"✓ Annotated task {tid} with '{LABEL_VALUE}'")
                done += 1
            except requests.HTTPError as e:
                print(f"! Failed on task {tid}: {e}")

    print(f"\nDone. Annotated {done} task(s).")

if __name__ == "__main__":

    # noinspection PyTypeChecker
    args = set(map(str.lower, sys.argv[1:]))

    dry = "dry" in args
    only_unannotated = "all" not in args


    main(start_id=5016, end_id=10032, only_unannotated=only_unannotated, dry_run=dry)
```
