## Le Stemming avec NLTK

```python
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer

# Exemple avec PorterStemmer
ps = PorterStemmer()
print(ps.stem("running"))
print(ps.stem("studies"))
print(ps.stem("flies"))

# Exemple avec SnowballStemmer (supporte plusieurs langues)
snow = SnowballStemmer("english")
print(snow.stem("running"))
```

## La Lemmatisation avec NLTK

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# Télécharger les ressources nécessaires (à faire une seule fois)
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

lemmatizer = WordNetLemmatizer()

# Sans préciser la catégorie grammaticale
print(lemmatizer.lemmatize("running"))   # running

# Avec la catégorie grammaticale (verbe)
print(lemmatizer.lemmatize("running", pos="v"))  # run
print(lemmatizer.lemmatize("studies", pos="n"))  # study
```

## Exemple comparatif

```python
words = ["running", "studies", "flies", "better"]

ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for w in words:
    print(f"{w:10} → Stem: {ps.stem(w):10} | Lemma (v): {lemmatizer.lemmatize(w, pos='v')}")
```

## Exemple complet — Stemming en français avec NLTK

```python
import nltk
from nltk.stem.snowball import SnowballStemmer

# Téléchargement des ressources de base
nltk.download('punkt')

# Texte d'exemple
texte = """
Les chercheurs en intelligence artificielle développent des systèmes capables
de comprendre, d'apprendre et de raisonner comme les humains.
Ils étudient également la traduction automatique et la reconnaissance vocale.
"""

# Initialisation du stemmer français
stemmer = SnowballStemmer("french")

# Tokenisation du texte
tokens = nltk.word_tokenize(texte, language="french")

# Application du stemming
racines = [stemmer.stem(token) for token in tokens]

# Affichage comparatif
for mot, racine in zip(tokens, racines):
    print(f"{mot:20} → {racine}")
```

> **Résumé**
> **Avantage :** rapide et efficace pour simplifier le texte avant le calcul de fréquences.
> **Limite :** perte de précision linguistique (*« intellig »* au lieu de *« intelligence »*).

## Exemple pratique : nettoyage de texte + stemming + TF-IDF

```python
import nltk
from nltk.stem.snowball import SnowballStemmer
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download('stopwords')
nltk.download('punkt')

stemmer = SnowballStemmer("french")
stop_words = stopwords.words('french')

corpus = [
    "Les systèmes intelligents apprennent à comprendre la parole humaine.",
    "La traduction automatique progresse grâce au deep learning.",
    "L'apprentissage supervisé est une méthode efficace en IA."
]

def preprocess(text):
    tokens = nltk.word_tokenize(text, language="french")
    stems = [stemmer.stem(t.lower()) for t in tokens if t.isalpha() and t.lower() not in stop_words]
    return " ".join(stems)

corpus_stemmed = [preprocess(t) for t in corpus]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus_stemmed)

print(vectorizer.get_feature_names_out())
print(tfidf_matrix.toarray())
```

## Bonus : pipeline hybride (NLTK + SpaCy)

```python
import nltk
import spacy
from nltk.corpus import stopwords
from nltk.stem.snowball import SnowballStemmer

nltk.download('punkt')
nltk.download('stopwords')

nlp = spacy.load("fr_core_news_md")
stemmer = SnowballStemmer("french")
stop_words = stopwords.words('french')

texte = "Les chercheurs développent des applications intelligentes pour la reconnaissance vocale."

# Étape 1 : tokenisation et nettoyage
tokens = [t for t in nltk.word_tokenize(texte, language="french") if t.isalpha()]
tokens = [t for t in tokens if t.lower() not in stop_words]

# Étape 2 : stemming + lemmatisation
for t in tokens:
    lemma = nlp(t)[0].lemma_
    stem = stemmer.stem(t)
    print(f"{t:15} → Lemma: {lemma:15} | Stem: {stem}")
```
